Regular Expressions (regex)

```python
import re

# Example using r for file paths
my_folder = r"C:\Exampledesktop\notes"
print(my_folder)

# re function re.search
result_search = re.search("pattern", r"searching in the string containing the pattern")
print(result_search)

# if the pattern doesn't exist
result_search_2 = re.search("nigger", r"No, niggas here")
print(result_search_2)

# re.sub
incorrect_strin = "The ninjas are black"
incorrect_strin_Corrected = re.sub("ninjas", "niggas", incorrect_strin)
print(incorrect_strin_Corrected)

#  r.?

customer_reviews = ["sam was a grat help to me in the store",
                    "the cashier was very rude to me, I think her name was eleasnor",
                    "amazing work from sadeen!",
                    "sarah was able to help me find the items I needed quickly", 
                    "The Sarah, is a very atrractive girl",
                    "great service from sara she found me what I wanted",
                    "SARAH's hair was in my food"]

sarah_reviews = []
pattern_to_find = r"sarah?"

for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        sarah_reviews.append(string)

        print(sarah_reviews)

# # finding the strings starting with a from the same customer_reviews (i.e ^ start of the string)
a_reviews = []
pattern_to_find = r"^a"
for string in customer_reviews:
    if(re.search(pattern_to_find, string)):
        a_reviews.append(string)

print(a_reviews)

# # finding the strings ending with y from the same customer_reviews (i.e $ end of string)
y_reviews = []
pattern_to_find = r"y$"
for string in customer_reviews:
    if(re.search(pattern_to_find, string)):
        y_reviews.append(string)

print(y_reviews)

# # finding with more than one string from the same customer_reviews (i.e | more than one)
needwant_reviews = []
pattern_to_find = r"(need|want)ed"

for string in customer_reviews:
    if(re.search(pattern_to_find, string)):
        needwant_reviews.append(string)

print(needwant_reviews)        

# remove panctuation from the strings
no_puct_reviews = []
pattern_to_find = r"[^\w\s]"

for string in customer_reviews:
    no_punct_strings = re.sub(pattern_to_find, "", string)
    no_puct_reviews.append(no_punct_strings) 

print(no_puct_reviews)
```

Tokenization: (NLTK punkt_tab)

```python
sentences = "Her cat's name is Luna. Her dog name is Tony"

print(sent_tokenize(sentences))
print(word_tokenize(sentences))
```

Stemming: PorterStemmer (NLTK)

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()

# example 1
connect_tokens = ['connecting', 'connected', 'connections', 'connectivity', 'connector', 'Connectify']
for tokens in connect_tokens:
    print(tokens, ": ", ps.stem(tokens))

# example 2
print (" ")
learn_tokens = ["learned", "learning", "learn", "learns", "learner", "learners"]
for token in learn_tokens:
    print(token, ":", ps.stem(token))

# example 3
print (" ")
likes_tokens = ['likes', 'better', 'worse']
for token in likes_tokens:
    print(token, ":", ps.stem(token))
 
```

Lemmatization

```python
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
lem = WordNetLemmatizer()

# applying them to stem example list 
for t in connect_tokens:
    print(t, ":", lem.lemmatize(t))

print (" ")
for token in learn_tokens:
    print(token, ":", lem.lemmatize(token))    


print (" ")
for token in likes_tokens:
    print(token, ":", lem.lemmatize(token))    
```
