lower()

```python
# String Lower:

sentence = "Her cat's name is Luna"

lower_scentence = sentence.lower()

print(lower_scentence)

# List Lower:
sentence_list = ['Could you pass me the TV remote?', 'It is IMPOSSIBLE to find this hotel', 'Want to go for dinner on Tuesday?']
lower_scentence_list = [x.lower() for x in sentence_list]
print(lower_scentence_list)
```
