```python
def monthly(rent):
    maintainance = 1500
    return rent + maintainance

def with_late(rent, date):
    new_days = date - 30
    return monthly(rent)*new_days + 10

date = 60
rent = 18000

if date > 30:
    print(with_late(rent, date))
else:
    print(monthly(rent))
```
