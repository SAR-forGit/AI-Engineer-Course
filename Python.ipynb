List Comprahensive

```python
numbers = [1, 13, 4, 5, 63, 100]
print(numbers)

new_numbers = []

for n in numbers:
    new_numbers.append(n*2)

print(new_numbers)

# same output in defferent format 
numbers = [2, 12, 22, 32, 42, 55]
new_numbers = [n*2 for n in numbers]
print(new_numbers)
```

```python
# Standard way 
print("Standard Way:", end=" ")
for i in range(2):
    for j in range(5):
        print(i + j, end=" ")

# Alternate Way (List Comprahensive)
new_list_comprahensive_1 = [i+j for i in range(2) for j in range(5)]
print("Altarnate Way:", new_list_comprahensive_1)
```

```python
# Another Way
new_list_comprahensive_2 = [[i+j for i in range(2)]for j in range(5)]
new_list_comprahensive_2
```

Lambda Function

```python
# Normal Function
def Square_giver(x):
    return x**2

Square_giver(3)

# Same thing using Lambda Function
Square_giver = lambda x: x**2
Square_giver(4)

# The no variable name Lambda Function 
(lambda x: x**2)(5)

# Lambda Function with 2 variables
(lambda x, y: x + y)(4, 5)

#  Nested Lambda Function
sum_xy = lambda x, y: x + y(x) # Here Its like saying Abdur rahman and Nowsh ka bada beta
sum_xy(2, lambda x:x*5)

# Lambda Function can have one or many parameters but can contain a single expression only
# Lambda Fucntion have same speed of regular function
```
